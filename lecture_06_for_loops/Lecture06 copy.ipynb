Now, with a **for loop**

````python
# Now, with a for loop

car_list = ["KIA", "Ferrari", "Ford", "Tesla"]

for car_name in car_list:
    print("Dear customer, we are writing about your " + car_name + " vehicle.")

# The code "for car_name in car_list:", means... 

# A variable "car_name" takes the value of every item in the list "car_list", one by one
# For each value of car_name, the indented block of code is executed once.
````

````python
# While Loop

success_count = 0
t = 0.0
final_time = 1.0
dt = 0.1

# while t != final_time:  # <--- WARNING: using this instead will cause infinite loop!

while abs(t - final_time) > 1e-13:
    # Experiment
    random_outcome = random.choice(possible_outcomes)

    # If it was a success, count it by increasing the success_count variable
    if random_outcome == "Success":
        success_count = success_count + 1

    # Increase time to the next value
    t = t + dt

print(success_count)

    
````

## <span style="color:red; font-size:25px"> III. Interrupting `for` loops </span>

<font size = "4">

- Sometimes we want to halt a "for" loop once a condition has been met. We can do so with the `break` keyword.

- Sometimes we want to skip over one pass through a "for" loop if some condition holds. We can do so with a `continue` keyword.

Suppose we have a collection of data stored in a list and we want to compute the sum of the elements. If there is no missing or corrupted data, its straightforward to do this with a for loop:

````python
data_list = [1, 2, 13, 4, -2, 11, 2.2]

data_sum = 0

for value in data_list:
    data_sum = data_sum + value 

print(data_sum)
````

<font size = "4">

But what if some of the data is missing/corrupted? We'll get an error in the next cell.

````python
type(None)
````

````python
data_list = [1, 2, None, 4, "missing", 11, 2.2]

data_sum = 0

for value in data_list:
    data_sum = data_sum + value 

print(data_sum)
````

<font size = "4">

**Option 1:** If we encounter a non-numerical data element, immediately stop the computation, warn the user and print what was computed so far. Use the `break` keyword.

````python
data_list = [1, 2, None, 4, "missing", 11, 2.2]

print(type(data_list[0]))   # valid data
print(type(data_list[6]))   # valid data
print(type(data_list[2]))   # invalid data
print(type(data_list[4]))   # invalid data
````

````python
data_list = [1, 2, None, 4, "missing", 11, 2.2]

data_sum = 0

for value in data_list:
    if (type(value) != int) and (type(value) != float):
        print("Warning: Non-numerical element in list! Exiting...")
        print("Value of sum at termination is:", data_sum)
        break #exit the loop
    data_sum = data_sum + value

print(data_sum)
````

<font size = "4">

**Option 2:** If we encounter a non-numerical data element, warn the user, but continue summing the rest of the list. Use the `continue` keyword.

````python
data_list = [1, 2, None, 4, "missing", 11, 2.2]

data_sum = 0

for value in data_list:
    if (type(value) != int) and (type(value) != float):
        print("Warning: Non-numerical element in list!")
        continue 
    data_sum = data_sum + value

print(data_sum)
````
